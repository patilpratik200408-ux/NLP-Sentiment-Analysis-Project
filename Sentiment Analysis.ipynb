```python
import pandas as pd
```

```python
df = pd.read_csv("../../Dataset/Restaurant_Reviews.tsv",sep="\t")
```

```python
df.head()
```

```python
x = df['Review']
```

```python
import re
from nltk.corpus import stopwords
from nltk import PorterStemmer
stemmer =PorterStemmer()
stopwords = stopwords.words('english')
stopwords.remove('not')
```

```python
def preprocessing(x):
    x = x.lower()
    pattern = '[^a-z]'
    x = re.sub(pattern," ",x)
    words = [stemmer.stem(word) for word in x.split() if word not in stopwords]
    return " ".join(words)
```

```python
prepro = df['Review'].apply(preprocessing)
```

```python
prepro
```

```python
from sklearn.feature_extraction.text import CountVectorizer,TfidfVectorizer
bow = TfidfVectorizer()
x_b = bow.fit_transform(prepro).toarray()
```

```python
y = df['Liked']
```

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression()
model.fit(x_b,y)
model.score(x_b,y)
```
